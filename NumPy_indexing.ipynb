```python
import numpy as np
```

```python
#Creating sample array
arr = np.arange(0,11)
```

```python
#Show
arr
```

```python
#Get a value at an index
arr[8]
```

```python
#Get values in a range
arr[1:5]
```

```python
#Get values in a range
arr[0:5]
```

```python
#Setting a value with index range (Broadcasting)
arr[0:5]=100

#Show
arr
```

```python
# Reset array, we'll see why I had to reset in  a moment
arr = np.arange(0,11)

#Show
arr
```

```python
#Important notes on Slices
slice_of_arr = arr[0:6]

#Show slice
slice_of_arr
```

```python
#Change Slice
slice_of_arr[:]=99

#Show Slice again
slice_of_arr
```

```python
arr
```

```python
#To get a copy, need to be explicit
arr_copy = arr.copy()

arr_copy
```

```python
arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))

#Show
arr_2d
```

```python
#Indexing row
arr_2d[1]
```

```python
# Format is arr_2d[row][col] or arr_2d[row,col]

# Getting individual element value
arr_2d[1][0]
```

```python
# Getting individual element value
arr_2d[1,0]
```

```python
# 2D array slicing

#Shape (2,2) from top right corner
arr_2d[:2,1:]
```

```python
#Shape bottom row
arr_2d[2]
```

```python
#Shape bottom row
arr_2d[2,:]
```

```python
#Set up matrix
arr2d = np.zeros((10,10))
```

```python
arr_length = len(arr2d)
for i in range(arr_length):
    arr2d[i] = i

arr2d
```

```python
#Length of array
arr_length = arr2d.shape[1]
```

```python
arr2d[[2,4,6,8]]
```

```python
#Allows in any order
arr2d[[6,4,2,7]]
```
